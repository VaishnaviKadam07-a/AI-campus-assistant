```python
import pandas as pd
import random

data = []

for i in range(1, 501):  # 500 students
    electricity = random.choice(["Low", "Medium", "High"])
    water = random.choice(["Yes", "No"])
    plastic = random.choice(["Low", "Medium", "High"])
    transport = random.choice(["Walk", "Bicycle", "Bus", "Bike"])
    recycling = random.choice(["Yes", "No"])

    # Eco Score Logic
    eco_score = 0
    eco_score += 3 if electricity == "Low" else 2 if electricity == "Medium" else 1
    eco_score += 2 if water == "Yes" else 0
    eco_score += 3 if plastic == "Low" else 2 if plastic == "Medium" else 1
    eco_score += 3 if transport in ["Walk", "Bicycle"] else 2 if transport == "Bus" else 1
    eco_score += 2 if recycling == "Yes" else 0

    data.append([
        f"S{i}",
        electricity,
        water,
        plastic,
        transport,
        recycling,
        eco_score
    ])

df_actions = pd.DataFrame(data, columns=[
    "student_id",
    "electricity_usage",
    "water_saving",
    "plastic_usage",
    "transport_mode",
    "recycling",
    "eco_score"
])

df_actions.to_csv("student_eco_actions.csv", index=False)

print("Student Eco Actions dataset created!")
```
